```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("train.csv")
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df[df['Postal Code'].isna()]
```

```python
df['Postal Code'] = df['Postal Code'].fillna(5401.0)
```

```python
df.isna().sum()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], format='%d/%m/%Y')

```

```python
df = df.drop(columns=['Row ID'])
```

```python
df.head()
```

```python
df['Order Year'] = df['Order Date'].dt.year
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days

df[['Order Date', 'Ship Date', 'Order Year', 'Shipping Days']].head()
```

```python
df['Shipping Days'].describe()
```

```python
category_sales = df.groupby('Category')['Sales'].sum().reset_index()
```

```python
category_sales = category_sales.sort_values(by='Sales', ascending=False)
```

```python
category_sales
```

```python
sub_category_sales = df.groupby('Sub-Category')['Sales'].sum().reset_index()
sub_category_sales = sub_category_sales.sort_values(by='Sales', ascending=False)
sub_category_sales
```

```python
segment_sales = df.groupby('Segment')['Sales'].sum().reset_index()
segment_sales = segment_sales.sort_values(by='Sales', ascending=False)
segment_sales['Percentage'] = (segment_sales['Sales']/segment_sales['Sales'].sum())*100
segment_sales['Percentage'] = segment_sales['Percentage'].round(2)
segment_sales
```

```python
region_sales = df.groupby('Region')['Sales'].sum().reset_index()
region_sales = region_sales.sort_values(by='Sales', ascending=False)
region_sales['Percentage'] = (region_sales['Sales']/region_sales['Sales'].sum())*100
region_sales['Percentage'] = region_sales['Percentage'].round(2)
region_sales
```

```python
top_products_sales = df.groupby('Product Name')['Sales'].sum().reset_index()
top_products_sales = top_products_sales.sort_values(by='Sales', ascending=False).head(10)
top_products_sales
```

```python
yearly_sales = df.groupby('Order Year')['Sales'].sum().reset_index()
yearly_sales = yearly_sales.sort_values(by='Order Year', ascending=True)
yearly_sales
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=yearly_sales, x='Order Year', y='Sales')

plt.title('Total Sales Trend by Year (2015-2018)')

plt.xlabel('Order Year')
plt.ylabel('Total Sales ($)')
plt.xticks(yearly_sales['Order Year'])

plt.show()
```

```python
yearly_category_sales = df.groupby(['Order Year', 'Category'])['Sales'].sum().reset_index()
yearly_category_sales = yearly_category_sales.sort_values(by=['Order Year','Sales'], ascending=(True, False))
yearly_category_sales
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=yearly_category_sales, x='Order Year', y='Sales', hue='Category')
plt.title('Sales by Category Per Year (2015-2018)')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=top_products_sales, x='Sales', y='Product Name')

plt.title('Top Selling Products')

plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=region_sales, x='Region', y='Percentage')
plt.title('Sales Distribution by Region (%)')
plt.ylabel('Percentage (%)')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=segment_sales, x='Segment', y='Sales')
plt.title('Sales by Customer Segment')
plt.xlabel('Customer Segment')
plt.ylabel('Total Sales ($)')
plt.show()
```

### 1. Revenue Trends & Growth Drivers (2015-2018)
* **Overall Trend:** Store sales experienced a slight dip in 2016 before rebounding strongly with continuous growth through 2017 and 2018.
* **Key Observations:** 
    * The 2016 dip was driven primarily by a decline in **Office Supplies** sales (dropping from $149.5K to $133.1K).
    * The recovery in 2017 was broad-based, but propelled mainly by **Technology**, which surged by ~37% (from $162.2K to $221.9K) and became the primary revenue driver.

### 2. Top Products Performance
* **Key Finding:** The **Canon Copier** is the top-selling product (over $60K)-more than double the sales of the second-ranked product.
* **Risk & Dependency:** The business has a high revenue concentration in a single product. any supplier disruption or drop in demand could severely hurt overall revenue.
* **Actionable Insight:** Maintain steady inventory levels for top-sellers while diversifying the product catalog to reduce reliance on the Canon copier.

### 3. Regional Performance & Customer Segments
* **Geographical Distribution:** The **West** and **East** regions drive over 60% of total sales, while the **South** represents the smallest market share (~17%).
* **Primary Customer Segment:** The **Consumer** segment is the main revenue driver, generating over $1.1M-exceeding Corporate and Home Office segments combined.
* **Actionable Insight:** Focus marketing efforts on sustaining leadership in the West and East, while exploring growth opportunities and localized campaigns in the South.
